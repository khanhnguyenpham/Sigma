```python
from pathlib import Path
import sys
import pandas as pd
candidates = [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]
project = next(p for p in candidates if (p / 'M2/models/model_families.json').is_file())
if str(project) not in sys.path:
    sys.path.insert(0, str(project))
from src.common import validate_run
run_id = 'm2_improved_models_20261006_fixed'
folder = project / 'M2/artifacts' / run_id
manifest = validate_run(folder)
assert manifest['status'] == 'complete'
```

```python
summary = pd.read_csv(folder / 'summary.csv')
display(summary[['family', 'target', 'mean_mape_pct', 'routes_le20', 'minimum_coverage']])
selected = pd.read_csv(folder / 'overall_selected_models.csv')
display(selected)
overall = pd.read_csv(folder / 'overall_test_metrics.csv')
primary = overall.loc[overall.block.eq(1) & overall.cadence.eq('daily_origins')]
display(primary.groupby('target').agg(mean_mape=('mape_positive_pct', 'mean'), routes_le20=('passes_20_pct', 'sum'), coverage=('coverage', 'min')))
```
